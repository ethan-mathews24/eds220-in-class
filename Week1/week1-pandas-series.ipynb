# Series

The integer number -999 is often used to represent missing values. Create a pandas.Series named s with four integer values, two of which are -999. The index of this series should be the letters A through D.

```python
s = pd.Series([1, 2, -999, -999], index = ['A', 'B', 'C', 'D'])
s
```

In the pandas.Series documentation, look for the method mask(). Use this method to update the series s so that the -999 values are replaced by NA values. HINT: check the first example in the method’s documentation

```python
s = s.mask(s == -999)
s
```

# DataFrames

```python
# Initialize dictionary with columns' data
d = {'col_name_1' : pd.Series(np.arange(3)),
    'col_name_2' : pd.Series([3.1, 3.2, 3.3])}

# Create the data frame

df = pd.DataFrame(d)
df
```

```python
# Change the index
df.index = ['a', 'b', 'c']
df
```
