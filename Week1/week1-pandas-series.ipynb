# DataFrames

```python
# Initialize dictionary with columns' data
d = {'col_name_1' : pd.Series(np.arange(3)),
    'col_name_2' : pd.Series([3.1, 3.2, 3.3])}

# Create the data frame

df = pd.DataFrame(d)
df
```

```python
# Change the index
df.index = ['a', 'b', 'c']
df
```

```python
# Change the column names
df.columns =['C1', 'C2']
df
```
